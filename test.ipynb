```python
#operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a+b
hasil2 = a-b
hasil3 = a*b
hasil4 = a/b
hasil5 = a**2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
#operator Elementwise ada masalah di perkalian matriks karena mengalikan per elemen
a = np.array( [(1,2,3), (4,5,6)] )
b = np.array( [(3,4,5), (3,4,5)] )

hasil = a*b
print("hasil =", hasil)
```

```python
# 3 Dimensi
from mpl_toolkits import mplot3d

fig = plt.figure()
ax = plt.axes(projection = '3d')

def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X,Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
#tugas2
import matplotlib.pyplot as plt

# Menyiapkan grid x dan y
x = np.linspace(-1.5, 1.5, 400)
y = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x, y)

# Definisi persamaan implisit
F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2

# Plot grafik implisit F(X, Y) = 1
plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, levels=[1], colors='red')
plt.title(r'Grafik 2D: $x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1$')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
```

```python
#tugas3
from mpl_toolkits import mplot3d

# Menyiapkan domain data x dan y
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)

# Definisi fungsi 3D
Z = (5*Y/4 - np.abs(X))**2 + X**2

# Plot permukaan 3D
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')

# Membuat plot permukaan dengan colormap
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

ax.set_title(r'Grafik 3D: $z = \left(\frac{5y}{4} - |x|\right)^2 + x^2$')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=5)

plt.show()
```
